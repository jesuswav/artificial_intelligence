## Perceptrón Simple AND

```python
import numpy as np

# Datos de entrada (x1, x2)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Salida deseada (AND)
y = np.array([0, 0, 0, 1])

# Hiperparámetros
lr = 0.1      # tasa de aprendizaje
epochs = 20   # número de iteraciones
np.random.seed(1)
w = np.random.randn(2)  # pesos iniciales
b = np.random.randn()   # sesgo

# Función de activación (escalón)
def step(x):
    return np.where(x >= 0, 1, 0)

# Entrenamiento
for epoch in range(epochs):
    for xi, yi in zip(X, y):
        z = np.dot(xi, w) + b
        y_pred = step(z)
        error = yi - y_pred
        # Regla de actualización
        w += lr * error * xi
        b += lr * error

print("Pesos finales:", w)
print("Bias final:", b)

# Prueba
for xi in X:
    print(f"{xi} → {step(np.dot(xi, w) + b)}")
```

## Perceptrón Multicapa XOR

```python
import numpy as np

# Datos de entrada (XOR)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([[0], [1], [1], [0]])

# Inicialización de pesos
np.random.seed(42)
w1 = np.random.randn(2, 3)  # pesos capa oculta (2 -> 3 neuronas)
b1 = np.zeros((1, 3))
w2 = np.random.randn(3, 1)  # pesos capa salida (3 -> 1)
b2 = np.zeros((1, 1))

# Función de activación y derivada (sigmoide)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Parámetros de entrenamiento
lr = 0.1
epochs = 10000

# Entrenamiento
for epoch in range(epochs):
    # --- Forward pass ---
    z1 = np.dot(X, w1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)

    # --- Cálculo del error ---
    error = y - a2
    mse = np.mean(np.square(error))

    # --- Backpropagation ---
    d_a2 = error * sigmoid_derivative(a2)
    d_a1 = np.dot(d_a2, w2.T) * sigmoid_derivative(a1)

    # --- Actualización de pesos ---
    w2 += lr * np.dot(a1.T, d_a2)
    b2 += lr * np.sum(d_a2, axis=0, keepdims=True)
    w1 += lr * np.dot(X.T, d_a1)
    b1 += lr * np.sum(d_a1, axis=0, keepdims=True)

    # Mostrar progreso cada 1000 iteraciones
    if epoch % 1000 == 0:
        print(f"Época {epoch}, Error: {mse:.4f}")

# --- Resultados finales ---
print("\nPredicciones finales:")
for xi, yi in zip(X, y):
    z1 = np.dot(xi, w1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)
    print(f"{xi} → {a2[0][0]:.4f}")
```
